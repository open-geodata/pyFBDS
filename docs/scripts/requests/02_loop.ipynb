```python
from pathlib import Path

import fbds
```

```python
project_path = Path('.').absolute().parents[1]

# Diretório de saída
output_path = project_path / 'data' / 'output'
output_path.mkdir(parents=True, exist_ok=True)

# Diretório de saída
log_path = project_path / 'data' / 'log'
log_path.mkdir(parents=True, exist_ok=True)
```

```python
fdbs = fbds.FBDS()
```

### Municipios

Lista municípios de um estado.


```python
muns = fdbs.get_municipalities(uf='SP')
muns
```

Obtem um município específico.


```python
mun = fdbs.get_municipalitie(uf='SP', municipality='SANTOS')
mun
```

<br>

---

### Layers

Lista layers de um município.


```python
# Inicializa o logger uma única vez antes do loop
logger = fbds.FBDSLogger(log_dir=log_path)
logger.start_download_session()


for muninicio in muns:
    municipio_name = muninicio["name"]
    print(municipio_name)

    lyrs = fdbs.get_layers(municipality=municipio_name, uf="SP")
    for lyr in lyrs:
        layer_name = lyr["name"]
        print(layer_name)

        # Acessa o layer
        lyr = fdbs.get_layer(municipality=municipio_name, uf="SP", layer=layer_name)

        # Lista de arquivos para download (do exemplo)
        files_to_download = fdbs.get_links(url=lyr["url"], ignore_first=2)
        # print(files_to_download)

        # Download usando threads (melhor para downloads)
        results_thread = fbds.download_files_parallel(
            url_list=files_to_download,
            output_dir=output_path,
            # Número de downloads simultâneos
            max_concurrent=4,
        )

# Finaliza a sessão após todo o processamento
logger.end_download_session()
```
