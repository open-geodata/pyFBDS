## Download


```python
project_path = Path('.').absolute().parents[2]
project_path
```

```python
# Diretório de saída
output_dir = project_path / 'data' / 'output'
output_dir.mkdir(parents=True, exist_ok=True)
output_dir
```

```python
# Lista de arquivos para download (do exemplo)
files_to_download = fdbs.get_links(url=layer['url'], ignore_first=2)
files_to_download[0:3]
```

```python
# Download usando threads (melhor para downloads)
print("Iniciando download com ThreadPoolExecutor...")

results_thread = pyFDBS.download_files_parallel(
    url_list=files_to_download,
    output_dir=output_dir,
    # Número de downloads simultâneos
    max_concurrent=4,
)
```

```python
pyFDBS.download_files_parallel
```

```python
# Download usando asyncio
print("Iniciando download assíncrono...")

try:
    results = pyFDBS.download_files_async(
        url_list=files_to_download,
        output_dir=output_dir,
        max_concurrent=5,  # Número de downloads simultâneos
    )
except:
    pass
```

```python
# Mostra resultados
print("Resultados:")
successes = [r for r in results_thread if r["status"] == "sucesso"]
errors = [r for r in results_thread if r["status"] == "erro"]
cached = [r for r in results_thread if r.get("cached", False)]


print(f"Downloads com sucesso: {len(successes)} de {len(results_thread)}")
print(f"Erros: {len(errors)} de {len(results_thread)}")
print(f"Arquivos do cache: {len(cached)} de {len(results_thread)}")

# Mostra erros se houver
if errors:
    print("Erros encontrados:")
    for error in errors:
        print(f"- {error['nome']}: {error['erro']}")
```
