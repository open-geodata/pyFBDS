## Download


```python
project_path = Path('.').absolute().parents[1]
project_path
```

```python
# Lista de arquivos para download (do exemplo)
files_to_download = fdbs.get_links(url=lyr['url'], ignore_first=2)
files_to_download
```

```python
# Diretório de saída
output_dir = project_path / 'data' / 'output'
output_dir.mkdir(parents=True, exist_ok=True)

# Download usando threads (melhor para downloads)
print("Iniciando download com ThreadPoolExecutor...")
results_thread = fbds.download_files_parallel(
    url_list=files_to_download,
    output_dir=output_dir,
    max_concurrent=4,  # Número de downloads simultâneos
    #use_threads=True,
)
```
