```python
# Lista de arquivos para download (do exemplo)
files_to_download = fdbs.get_links(url=layer["url"], ignore_first=2)

# Diretório de saída
# output_dir = Path('downloads/ADAMANTINA/HIDROGRAFIA')

# Download usando asyncio
print("Iniciando download assíncrono...")
results = pyFDBS.download_files_async(
    url_list=files_to_download,
    output_dir=output_dir,
    max_concurrent=5,  # Número de downloads simultâneos
)

# Mostra resultados
print("\nResultados:")
successes = [r for r in results if r["status"] == "sucesso"]
errors = [r for r in results if r["status"] == "erro"]

print(f"Total de arquivos: {len(results)}")
print(f"Downloads com sucesso: {len(successes)}")
print(f"Erros: {len(errors)}")

# Mostra erros se houver
if errors:
    print("\nErros encontrados:")
    for error in errors:
        print(f"- {error['nome']}: {error['erro']}")
```
