```python
import time
from pathlib import Path

import pandas as pd

from pyFBDS.compare import ComparadorTexto
from pyFBDS.logger import FBDSLogger
from pyFBDS.lxml import FBDS
```

```python
# Importa pacote
from pyFBDS.post import FBDS

# Instancia FBDS
fbds = FBDS(output_path='.')

# Faz Download dos Dados para output_path
fbds.download(municipality='SANTOS', uf='SP')

# Lê dados espaciais em formato geodataframe
gdf = fbds.read_data(municipality='SANTOS', uf='SP', layer="APP_USO")
```

## Pastas


```python
project_path = Path(".").absolute().parents[2]

# Diretório de saída
data_path = project_path / "data"

logs_path = data_path / "log"
logs_path.mkdir(parents=True, exist_ok=True)

output_path = data_path / "output"
output_path.mkdir(parents=True, exist_ok=True)
output_path
```

<br>

---

## FBDS

Instancia a classe


```python
# Instancia Logger
logger = FBDSLogger(
    log_dir=logs_path,
    new_session=True,
)

# Instancia FBDS
fdbs = FBDS(
    temp_path=output_path,
    logger=logger,
)
```

```python
municipality = 'SAO_PAULO2'
```

```python
# Compara texto
comparador = ComparadorTexto(limite_minimo=0.1)

# Retorna os X mais semelhantes
top_x = comparador.buscar_top_x_semelhantes(
    alvo=municipality,
    opcoes=fdbs.municipios(uf="SP"),
    top_x=15,
)
top_x
```

```python
# Filtrando os itens com score > 0.95
filtrados = [item for item in top_x if item[1] > 0.95]
filtrados

print(filtrados)
```
