```python
from pathlib import Path

from pyFBDS.logger import FBDSLogger
from pyFBDS.lxml import FBDS as FBDS_xml
from pyFBDS.post import FBDS, IBGE
```

<br>

---

## Pastas


```python
project_path = Path('.').absolute().parents[2]

# Diretório de saída
data_path = project_path / 'data'

logs_path = data_path / 'log'
logs_path.mkdir(parents=True, exist_ok=True)

output_path = data_path / 'output'
output_path.mkdir(parents=True, exist_ok=True)
output_path
```

<br>

---

## FBDS

Instancia a classe


```python
# Instancia Logger
# Inicializa o logger uma única vez antes do loop
logger = FBDSLogger(
    log_dir=logs_path,
    new_session=True,
    console=True,
)

# Instancia FBDS do módulo LXML
fbds = FBDS_xml(
    temp_path=output_path,
    output_path=output_path,
    logger=logger,
)
```

```python
ibge = IBGE(
    temp_path=output_path,
    fbds=fbds,
)

# Lê tabela
ibge.data.head()
```

```python
codigo_desejado = 3550308  # São Paulo - SP
# codigo_desejado = 3548500  # Santos - SP
codigo_desejado = 3545803  # Santa Bárbara d'Oeste - SP
#codigo_desejado = 3520400

municipio, uf = ibge.search(id_ibge=codigo_desejado)
```

<br>

---

## _Download_


```python
# Instancia FBDS
fbds = FBDS(
    temp_path=output_path,
    output_path=output_path,
    logger=logger,
)
```

```python
filepath = fbds.download(
    uf=uf,
    municipality=municipio,
)
```

<br>

---

## Lista Arquivos


```python
list_shps = fbds.list_shapefiles(filepath=filepath)
list_shps
```

```python
list_shps = fbds.clean_list(filepath=filepath)
list_shps
```

## Read

Ao invés de fazer _download_ do arquivo `.tar`, descompactar o arquivo etc, optei por criar uma função que me dê o layer em _runtime_, ou seja, em tempo de execução.

Dessa maneira torna-se mais fácil consumir os dados, trata-los etc.


```python
# Parâmetro
layer = "MASSAS_DAGUA"
#layer = "APP_USO"
layer = "APP"

# Cria geodataframe
gdf = fbds.read_data(
    uf=uf,
    municipality=municipio,
    layer=layer,
)
```

```python
gdf.explore(
    tiles="OpenStreetMap",
    attr='&copy; <a href="https://www.openstreetmap.org/copyright">OpenStreetMap</a> contributors &copy; <a href="https://carto.com/attributions">CARTO</a>',
)
```
