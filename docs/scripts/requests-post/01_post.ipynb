```python
from urllib.parse import parse_qs, parse_qsl, unquote, unquote_plus, urlencode, urlsplit

import requests

from fbds.scraper import webdriver
```

```python
aa = 'action=download&as=ACRELANDIA.tar&type=php-tar&baseHref=%2FAC%2F&hrefs=&hrefs%5B0%5D=%2FAC%2FACRELANDIA%2F'
urlsplit(unquote(aa))
```

```python
# A função parse_qs decompõe a string e decodifica os valores (como %2F para /)
# O parâmetro 'keep_blank_values=True' garante que 'hrefs=' seja mantido.
# O parâmetro 'encoding="utf-8"' é uma boa prática.
dados_decompostos = parse_qs(aa, keep_blank_values=True, encoding="utf-8")
print(dados_decompostos)
```

```python
params = dict(parse_qsl(aa))
params['hrefs'] = ''
params['action'] = '"download"'
params
```

```python
encode_params = urlencode(params)
#print('action=download&as=ASSIS_BRASIL.tar&type=php-tar&baseHref=%2FAC%2F&hrefs=&hrefs%5B0%5D=%2FAC%2FASSIS_BRASIL%2F')
print(encode_params)
```

```python
# 1. Defina seus Request Headers
request_headers = {
    #'Authorization': 'Bearer 12345abcdef',
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/142.0.0.0 Safari/537.36',
    # Se usar o parâmetro 'json=payload', o requests define Content-Type
    'Content-Type': 'application/x-www-form-urlencoded',
    # Se usar o parâmetro 'data=payload', o requests define Content-Type
    # Se passar string com 'data=', é bom definir o Content-Type:
    # 'Content-Type': 'application/json'
    #'Cookie': 'PHPSESSID=ce49648c5246824ee42b245661a412e5'
}
```

```python
driver = webdriver.Firefox(verify_ssl=False)
driver.get(url='https://geo.fbds.org.br/SP/')
```

```python
# get the URL from Selenium
current_pdf_url = driver.current_url

# create a requests session
session = requests.session()

# add Selenium's cookies to requests
selenium_cookies = driver.get_cookies()
for cookie in selenium_cookies:
    session.cookies.set(cookie['name'], cookie['value'])

# Note: If headers are also important, you'll need to use
# something like seleniumwire to get the headers from Selenium

# Finally, re-send the request with requests.session
pdf_response = session.get(current_pdf_url)
```

```python
r = requests.post(
    url="https://geo.fbds.org.br/AC/?",
    data=encode_params,
    headers=request_headers,
)
r.content.decode()
```

```python
# Verifica se a requisição foi bem-sucedida (código 200, 201, 202, etc.)
r.raise_for_status() 

# Se a API retornar JSON, você pode acessar os dados da resposta:
dados_resposta = r.json()
```
